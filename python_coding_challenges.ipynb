```python
# Two sum problem using brute force

num = [1,4,5,8,9,2,0,3]
target = 14

for i in range(len(num)):   #n
  for j in range(i+1, len(num)):   #n
    if num[i]+num[j]==target:
      print(f"Indices are {i} and {j}")
      print(f"values are {num[i]} and {num[j]}")
      break

#total time complixity will be O(n^2)
```

```python
# optimal approch of above problem

num = [1,4,5,8,9,2,0,3]
target = 14
seen={}

for i,v in enumerate(num):    #n
  comp=target-v
  if comp in seen:
    print(f"Indices are {i} and {seen[comp]}")
    print(f"values are {comp} and {v}")
    break
  else:
    seen[v]=i

# Time and space complixity of this program will be O(n)
```

```python
# reverse only the vowels

class Vowels:
    def reverseVowels(self, s:str) -> str:
        l=['a','e','i','o','u','A','E','I','O','U']
        left,right=0,len(s)-1
        s=list(s)

        while left < right:
            if s[left] not in l:
                left +=1

            elif s[right] not in l:
                right-=1
            else:
                s[left], s[right] = s[right], s[left]
                left+=1
                right-=1
        
        return ''.join(s)

v=Vowels()
print(v.reverseVowels("IceCream"))
```

```python
# Creating a stack    (LIFO)

class Stack:
  def __init__(self):
    self.stack=[]

  def push(self,elem):
    self.stack.append(elem)

  def pop(self):
    if self.isEmpty():
      return "Stack is empty"
    return self.stack.pop()

  def peek(self):
    if self.isEmpty():
      return "Stack does not contain any element"
    return self.stack[-1]

  def isEmpty(self):
    return len(self.stack)==0

  def size(self):
    return len(self.stack)

s=Stack()
s.push("A")
s.push("B")
s.push("C")
s.push("D")

print(f"size of stack: {s.size()}")
print(f"pop: {s.pop()}")
print(f"size of stack: {s.size()}")
print(f"peek: {s.peek()}")
print(f"pop: {s.pop()}")
print(f"pop: {s.pop()}")
print(f"is Empty: {s.isEmpty()}")
print(f"pop: {s.pop()}")
print(f"is Empty: {s.isEmpty()}")
print(f"size of stack: {s.size()}")
print(f"pop: {s.pop()}")
print(f"peek: {s.peek()}")
```

```python
# Queue   (FIFO)

class Queue:

  def __init__(self):
    self.queue=[]

  def enqueue(self, element):
    self.queue.append(element)

  def dequeue(self):
    if self.isEmpty():
      return "no element in queue"
    return self.queue.pop(0)

  def peek(self):
    if self.isEmpty():
      return "no element"
    return self.queue[0]

  def isEmpty(self):
    return len(self.queue) == 0

  def size(self):
    return len(self.queue)


q=Queue()
q.enqueue("a")
q.enqueue("b")
q.enqueue("c")
q.enqueue("d")
q.enqueue("e")

print(f"Size of queue: {q.size()}")
print(f"deque: {q.dequeue()}")
print(f"deque: {q.dequeue()}")
print(f"peek: {q.peek()}")
print(f"Size of queue: {q.size()}")
print(f"empty: {q.isEmpty()}")
```

Longest Palindromic Substring

```python
class Solution:
    def longestPalindrome(self, s: str) -> str:
      prev=""

      for i in range(len(s)):
        left,right=i,i

        # Odd length String
        while s[left] >= 0 and s[right] <= len(s) and s[left] == s[right]:
          if right-left+1 > len(prev):
            prev=s[left:right+1]

          left-=1
          right+=1

        # Even length String
        left=i,right=i+1
        while s[left] >= 0 and s[right] <= len(s) and s[left] == s[right]:
          if right-left+1 > len(prev):
            prev=s[left:right+1]

      return prev
l=Solution()
print(l.longestPalindrome("cbbbd"))
```

Longest Substring without repeating characters

```python
class Substring:
  def longestSubstring(self, s: str) -> str:
    seen=set()
    left=0
    max_length=0

    for right in range(len(s)):
      while s[right] in seen:
        seen.remove(s[left])
        left+=1

      seen.add(s[right])

      current_length=right - left + 1

      if current_length > max_length:
        max_length = current_length

    return max_length

s=Substring()
print(s.longestSubstring("pwwkew"))
print(s.longestSubstring("abcabcbb"))
```

```python
# Check the giving integer is palindrom or not without converting the integer into string

class CheckPalindrome:
  def isPalindrome(self, num: int) -> bool:
    rev = 0
    if num < 0 :
      return False
    value = num

    while num > 0 :
      rem = num % 10
      rev = rev * 10 + rem
      num = num // 10

    if value == rev:
      return True

p = CheckPalindrome()
print(p.isPalindrome(-121))
print(p.isPalindrome(121))
```

```python
# Reverse the given integer

class Reverse:
  def reverseInteger(self, num : int) -> int:
    rev=0
    sign = -1 if num < 0 else 1
    num=abs(num)

    while num > 0:
      rem = num % 10
      rev = rev * 10 + rem
      num = num // 10

    rev = rev * sign

    return rev

r=Reverse()
print(r.reverseInteger(-123))
print(r.reverseInteger(4232))
```

String to integer (myatoi)

```python
class Solution:
  def myAtoi(self, s: str) -> int:
    s=s.lstrip()

    if not s:
      return 0

    sign = 1
    i=0

    if s[0] == "-":
      sign = -1
      i+=1
    elif s[0] == "+":
      i+=1

    num = 0
    while i < len(s) and s[i].isdigit():
      num = num * 10 + int(s[i])
      i+=1

    num*=sign

    min_bound,max_bound=-(2**31),(2**31-1)

    if num < min_bound:
      return min_bound
    elif num > max_bound:
      return max_bound

    return num

s=Solution()
print(s.myAtoi("42"))
print(s.myAtoi("words and 987"))
print(s.myAtoi(" -042"))
print(s.myAtoi("1337c0d3"))
print(s.myAtoi("0-1"))
```

```python
# Reverse the string without using extra space inplace replacement

class Reverse:
    def reverseString(self, s : list[str]) -> list:
        left,right=0,len(s)-1

        while left < right:
            s[left],s[right]=s[right],s[left]
            left+=1
            right-=1
        return s

r=Reverse()
print(r.reverseString(["h","e","l","l","o"]))
```

```python
# Romam to integer    time-> O(n)   space-> O(1)

class Roman:
    def romanToInt(self, s : str) -> int:
        d = {"I":1, "V":5, "X":10, "L":50, "C":100, "M":1000}
        total = 0

        for ind, val in enumerate(s):
            if ind == len(s)-1:
                total += d[val]
            elif d[val] < d[s[ind + 1]]:
                total -= d[val]
            elif d[val] > d[s[ind + 1]]:
                total += d[val]
            else:
                total += d[val]

        return total

r=Roman()
print(r.romanToInt("III"))
print(r.romanToInt("LVIII"))
print(r.romanToInt("MCMXCIV"))
```

```python
# Valid paranthesis

class Paranthesis:
    def isValid(self, s: str) -> bool:
        stack = []
        my_dict = {"]":"[", "}":"{", ")":"("}

        for i in s:
            if i in my_dict.values():
                stack.append(i)
            else:
                if not stack or stack[-1] != my_dict[i]:
                    return False
                stack.pop()

        return len(stack) == 0

p=Paranthesis()
print(p.isValid("()"))   
print(p.isValid(""))    
print(p.isValid("(){}[]"))    
print(p.isValid("([{{}}])"))    
print(p.isValid("[))]"))
```
