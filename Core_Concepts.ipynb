Higher order functions: Higher order functions are the functions that takes the another function as an argument and returns the result.

example: map(), filter(), reduce(), starmap() and filterfalse()

```python
# map
num=[1,2,3,4,5,6,7,8,9]
square=list(map(lambda x:x*x,num))
print(square)
```

```python
# filter
num=[1,2,3,4,5,6,7,8,9]
eve=list(filter(lambda x:x%2==0,num))
print(eve)
```

```python
# reduce
from functools import reduce

nums=[1,5,8,9]
sum=reduce(lambda x,y:x+y,nums)
print(sum)
```

```python
# filterfalse
from itertools import filterfalse

num=[1,2,3,4,5,6,7,8,9]
square=list(filterfalse(lambda x:x%2==0,num))
print(square)
```

```python
# starmap
from itertools import starmap

nums=[[1,2],[3,4],[5,6]]
sum=list(starmap(lambda x,y:x+y,nums))
print(sum)
```

Scope of variables

global, local and nonlocal

```python
def print_name():
  name='Nikita'  # local variable
  return name

print_name()
```

```python
x=10

def greet():
  x=20 #local variable

  return x

greet()
```

```python
x=10

def greet():
  global x
  x=20 #local variable

  return x

greet()
```

```python
def outer():
  x=10

  def inner():
    x=20
    return x
  inner()
  return x

outer()
```

```python
def outer():
  x=10

  def inner():
    nonlocal x
    x=20
  inner()
  return x

outer()
```

```python
x='global'

def outer():
  x='enclosing'
  def inner():
    nonlocal x
    x='modified'
  inner()
  print(x)

outer()
```

Decorators

```python
# Static Method

class calculate:
  @staticmethod
  def add(a,b):
    return a+b
print(calculate.add(10,20))
```

```python
# class method

class Student:
  name="anonymous"
  @classmethod
  def setter(cls,name,age):
    cls.name=name
    cls.age=age
    return f"your name is {name} and age is {age}"

s=Student
print(s.setter('Nikita','23'))
```

```python
# propery

class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height

    @property
    def area(self):
        return self.width * self.height

r = Rectangle(5, 4)
print(r.area)
```

Magic Methods (dunder method)


__init__() runs automatically when an object is created, and __str__() runs automatically when the object is printed.

```python

class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"Name: {self.name}, Age: {self.age}"

s = Student("Nikita", 23)

print(s)
```

```python

class Marks:
    def __init__(self, score):
        self.score = score

    def __add__(self, other):
        return self.score + other.score

m1 = Marks(80)
m2 = Marks(90)

print(m1 + m2)
```

```python

class Student:
    def __init__(self, name):
        self.name = name

    def __eq__(self, other):
        return self.name == other.name

s1 = Student("Nikita")
s2 = Student("Nikita")

print(s1 == s2)
```

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

t = Team(["Nikita", "Aman", "Priya"])

print(len(t))
```

```python
class Greeting:
    def __call__(self, name):
        print(f"Hello, {name}!")

greet = Greeting()
greet("Nikita")
```

__iter__()

Returns an iterator object. It is called when you use iter() or begin a for loop over an iterable.

__next__()

Returns the next item from an iterator. When no items remain, it raises StopIteration.

```python
try:
  iterator=[10,20,30]

  it=iter(iterator) # calls __iter__()

  print(next(it)) #calls __next__()
  print(next(it))
  print(next(it))
  print(next(it))
except Exception as e:
  print(e)
```

Python __slots__ and __dict__

```python
class Demo:
  def __init__(self,name,age):
    self.name=name;
    self.age=age;

d=Demo("Nikita", 23)
print(d.__dict__)
```

```python
class Student:
  def __init__(self,name):
    self.name=name;

s=Student("nikita")
s.age=23
print(s.__dict__)
```

```python
class Student:
  __slots__=("name","age")
  def __init__(self,name,age):
    self.name=name
    self.age=age

s=Student("nikita",23)
# print(s.__dict__)  # Attrribute Error

print(s.name)
print(s.age)
```
