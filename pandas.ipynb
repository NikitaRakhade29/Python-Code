```python
# Handling missing data in the dataframe
import numpy as np

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit", "Sneha"],
    "Age": [25, np.nan, 30, 26],
    "Salary": [30000, 45000, np.nan, 35000]
})

print(df)
```

```python
print(df.isnull())
```

```python
print(df.isnull().sum())
```

```python
print(df.isna())
```

```python
print(df.notnull().sum())
```

```python
result = df.dropna() # this will drop the rows with missing values
# This will remove entire record of the person where na is present
print(result)
```

```python
res = df.fillna(0)  # fill the na values with 0
print(res)
```

```python
# forward and backward fill

forward_result = df.ffill()
print(forward_result)
```

```python
backward_result = df.bfill()
print(backward_result)
```

```python
# Remove duplicates 

df = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Rahul", "Amit"],
    "Age": [25, 28, 25, 30]
})

print(df.duplicated())  # identify duplicate rows
```

```python
res = df.drop_duplicates()
print(res)  # removes the rows with duplicates
```

```python
res1 = df.drop_duplicates(keep='first')  # keep only first duplicate row and remove others
print(res1)
```

```python
res2 = df.drop_duplicates(keep="last")  # keep the last occured duplicate and remove others 
print(res2)
```

```python
res3 = df.drop_duplicates(keep=False)  # remove all the duplicates in the dataframe
print(res3)
```

```python
df3 = pd.DataFrame({
    "Name": ["Rahul", "Priya", "Amit", "Sneha", "Karan"],
    "Department": ["IT", "HR", "IT", "HR", "Finance"],
    "Salary": [30000, 45000, 50000, 35000, 60000]
})
```

```python
data = df3.loc[df3["Salary"] > 30000].groupby("Department")["Salary"].mean().reset_index()
print(data)
```

```python
df3["Department_Avg"] = (
    df3.groupby("Department")["Salary"]
      .transform("mean")
)

print(df3)
```
