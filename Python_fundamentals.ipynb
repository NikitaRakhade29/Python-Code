SHALLOW COPY : A shallow copy creates a new outer object but does not recursively copy the objects inside it.

A reference is a way to access an object. Python variables are names bound to objects, rather than boxes that independently contain their values.

```python
a=[1,2,3,4,[5,6]]
b=a.copy() # SHALLOW COPY
c=a   # REFERENCE
b.append(8)
c.append(10)
print(b)
print(a)
print(c)
a[4].append(99)
print(a)
print(b)
print(a)
```

DEEP COPY : A deep copy creates a new object and recursively copies nested objects, so mutable nested structures are generally independent.

```python
import copy
a=[1,2,[3,4,5]]
b=copy.deepcopy(a)  # DEEP COPY

b[2].append(99)
print(a)
print(b)
```

```python
# Two sums problem (with indices and values)
nums = [2, 7, 11, 15]
target = 9

for i in range(len(nums)):
  find=target-nums[i]
  for j in range(i+1,len(nums)):
    if nums[j]==find:
      print(f'indices are: {i} {j}')
      print(f'The values are {nums[i]} and {nums[j]}')
      break
```

```python
# Group Anagrams
def group_anagrams(strings):
  anagram={}
  for word in strings:
    key="".join(sorted(word))
    if key not in anagram:
      anagram[key]=[]
    anagram[key].append(word)
  return anagram.values()

strings = ["eat", "tea", "tan", "ate", "nat", "bat"]
# Sample output should look like this [["eat", "tea", "ate"], ["tan", "nat"], ["bat"]]
print(group_anagrams(strings))
```

```python
# Top K Frequent Elements

nums = [1, 1, 1, 2, 2, 3]
k = 2
l = []
c = {}

for i in nums:
    c[i] = c.get(i, 0) + 1
# Sort by frequency (highest first),
# then by number (smallest first)
for key in sorted(c, key=lambda x: (-c[x], x)):
    l.append(key)
    if len(l) == k:
        break

print(l)
```

```python
# Longest Consecutive Sequence
nums = [100, 4, 200, 1, 3, 2]
s = set(nums)
longest = 0
for i in s:
    count = 1
    while i + count in s:
        count += 1
    if count > longest:
        longest = count
print(longest)
```

```python
# Subarray Sum Equals K
nums = [1, 2, 3]
k = 3
count = 0
for i in range(len(nums)):
    total = 0
    for j in range(i, len(nums)):
        total += nums[j]
        if total == k:
            count += 1
print(count)
```
